```python
import pandas as pd

df = pd.read_csv("gia_nha.csv")

nho = df.iloc[[0, 14, 29, 44, 59]]

x = nho["dien_tich"].to_numpy()
y = nho["gia"].to_numpy()


def mse(w, b):
    """Sai số trung bình bình phương của đường thẳng y = wx + b."""
    y_du_doan = w * x + b
    sai_so = y - y_du_doan
    return (sai_so ** 2).mean()


print("Năm căn được chọn:")

for i in range(5):
    print(f"  x = {x[i]:6.1f} m2  ->  y = {y[i]:5.2f} tỷ")


print()

for w, b in [(0.05, 1.5), (0.08, 0.5)]:
    print(f"Đường thẳng y = {w}x + {b}")

    for i in range(5):
        y_du_doan = w * x[i] + b
        sai_so = y[i] - y_du_doan

        print(
            f"  x = {x[i]:6.1f}  "
            f"thật = {y[i]:5.2f}  "
            f"dự đoán = {y_du_doan:5.2f}  "
            f"sai số = {sai_so:6.2f}"
        )

    print(f"MSE = {mse(w, b):.4f}")
    print()


print("Đường nào có MSE nhỏ hơn thì dự đoán dữ liệu tốt hơn.")
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv("gia_nha.csv")

x = df["dien_tich"].to_numpy()
y = df["gia"].to_numpy()

x_tb = x.mean()
y_tb = y.mean()

tu_so = ((x - x_tb) * (y - y_tb)).sum()
mau_so = ((x - x_tb) ** 2).sum()

w = tu_so / mau_so
b = y_tb - w * x_tb

print(f"Trung bình diện tích : {x_tb:.4f}")
print(f"Trung bình giá       : {y_tb:.4f}")
print(f"Tử số                : {tu_so:.4f}")
print(f"Mẫu số               : {mau_so:.4f}")
print()

print(f"Hệ số góc   w = {w:.6f}")
print(f"Hệ số chặn  b = {b:.6f}")
print()

print(f"Mô hình: gia = {w:.4f} * dien_tich + {b:.4f}")
print()

y_du_doan = w * x + b
mse = ((y - y_du_doan) ** 2).mean()

print(f"MSE trên toàn bộ 60 căn: {mse:.4f}")

print(f"Căn 80 m2 được dự đoán: {w * 80 + b:.3f} tỷ đồng")
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv("gia_nha.csv")

x_goc = df["dien_tich"].to_numpy()
y = df["gia"].to_numpy()

x_tb = x_goc.mean()
x_do_lech = x_goc.std()
x = (x_goc - x_tb) / x_do_lech

w, b = 0.0, 0.0

toc_do_hoc = 0.1
so_vong = 200
n = len(x)

print("Vong     w          b          MSE")

for vong in range(1, so_vong + 1):

    y_du_doan = w * x + b

    chenh_lech = y_du_doan - y

    grad_w = (2 / n) * (chenh_lech * x).sum()
    grad_b = (2 / n) * chenh_lech.sum()

    w -= toc_do_hoc * grad_w
    b -= toc_do_hoc * grad_b

    if vong in (1, 2, 5, 10, 25, 50, 100, 200):

        mse = ((w * x + b - y) ** 2).mean()

        print(
            f"{vong:4d}   "
            f"{w:7.4f}   "
            f"{b:7.4f}   "
            f"{mse:8.4f}"
        )

print()

w_goc = w / x_do_lech
b_goc = b - w * x_tb / x_do_lech

print("Sau khi đổi về thang đo mét vuông:")
print(f"w = {w_goc:.6f}")
print(f"b = {b_goc:.6f}")

print()
print("So với công thức ở bước 3: w = 0.078367, b = 0.401752")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

# 1. Đọc dữ liệu và chia tập học
df = pd.read_csv("gia_nha.csv")

x = df["dien_tich"].to_numpy()
y = df["gia"].to_numpy()

x_train, _, y_train, _ = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# Chuẩn hóa dữ liệu (Feature Scaling) để Gradient Descent hội tụ nhanh
x_mean, x_std = x_train.mean(), x_train.std()
x_train_scaled = (x_train - x_mean) / x_std

# 2. Huấn luyện bằng Gradient Descent
epochs = 200
lr = 0.1  # Learning rate

w = 0.0
b = 0.0
mse_history = []

n = len(x_train_scaled)

for epoch in range(epochs):
    y_pred = w * x_train_scaled + b

    # Tính MSE (Mean Squared Error)
    mse = np.mean((y_train - y_pred) ** 2)
    mse_history.append(mse)

    # Tính đạo hàm
    dw = (-2 / n) * np.sum(x_train_scaled * (y_train - y_pred))
    db = (-2 / n) * np.sum(y_train - y_pred)

    # Cập nhật w và b
    w -= lr * dw
    b -= lr * db

# 3. Vẽ Hình 10
plt.figure(figsize=(7, 4.5))

plt.plot(range(epochs), mse_history, color="#0B2545", linewidth=2.5)

# Cấu hình thang đo logarit cho trục y
plt.yscale("log")

# Đặt chú thích (Annotation) chỉ vào vị trí đứng yên
plt.annotate(
    "gần như không giảm thêm",
    xy=(120, mse_history[120]),
    xytext=(90, mse_history[120] * 3),
    arrowprops=dict(arrowstyle="->", color="#C02228", lw=1.5),
    color="#C02228",
    fontweight="bold",
    fontsize=10
)

# Cấu hình tiêu đề và nhãn trục
plt.title("Sai số giảm nhanh rồi đứng yên", fontsize=13, fontweight="bold", pad=10)
plt.xlabel("Vòng lặp", fontsize=11, fontweight="bold")
plt.ylabel("MSE (thang loga)", fontsize=11, fontweight="bold")

plt.grid(True, alpha=0.3, which="both", linestyle="-")
plt.tight_layout()

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv("gia_nha.csv")

x = df["dien_tich"].to_numpy()
y = df["gia"].to_numpy()

x_train, _, y_train, _ = train_test_split(
    x, y, test_size=0.2, random_state=42
)

x_mean, x_std = x_train.mean(), x_train.std()
x_train_scaled = (x_train - x_mean) / x_std

epochs = 61
lr = 0.1

w = 0.0
b = 0.0
n = len(x_train_scaled)

target_epochs = {0: "Lúc bắt đầu, chưa chạy vòng nào",
                 2: "Sau 2 vòng",
                 5: "Sau 5 vòng",
                 15: "Sau 15 vòng",
                 60: "Sau 60 vòng"}

lines_to_plot = {}

for epoch in range(epochs):
    if epoch in target_epochs:
        lines_to_plot[epoch] = (w, b)

    y_pred = w * x_train_scaled + b
    dw = (-2 / n) * np.sum(x_train_scaled * (y_train - y_pred))
    db = (-2 / n) * np.sum(y_train - y_pred)

    w -= lr * dw
    b -= lr * db

plt.figure(figsize=(7, 4.5))

plt.scatter(x_train, y_train, color="#3868A0", alpha=0.5, s=20, zorder=2)

x_line = np.linspace(x_train.min(), x_train.max(), 200)
x_line_scaled = (x_line - x_mean) / x_std

colors = {
    0: "#4A4A4A",   # Xám
    2: "#C28E0E",   # Vàng đất
    5: "#D9531E",   # Cam
    15: "#0E7C8B",  # Xanh lam
    60: "#C02228"   # Đỏ
}

for epoch, label in target_epochs.items():
    w_ep, b_ep = lines_to_plot[epoch]
    y_line = w_ep * x_line_scaled + b_ep
    plt.plot(x_line, y_line, color=colors[epoch], linewidth=2, label=label)

plt.title("Đường thẳng tự dịch dần về đúng chỗ", fontsize=13, fontweight="bold", pad=10)
plt.xlabel("Diện tích (m²)", fontsize=11, fontweight="bold")
plt.ylabel("Giá (tỷ đồng)", fontsize=11, fontweight="bold")

plt.legend(loc="upper left", fontsize=9)
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv("gia_nha.csv")

x = df["dien_tich"].to_numpy()
y = df["gia"].to_numpy()

x_train, _, y_train, _ = train_test_split(
    x, y, test_size=0.2, random_state=42
)

x_mean, x_std = x_train.mean(), x_train.std()
x_train_scaled = (x_train - x_mean) / x_std

def train_gd(lr, epochs=61):
    w, b = 0.0, 0.0
    n = len(x_train_scaled)
    mse_history = []

    for epoch in range(epochs):
        y_pred = w * x_train_scaled + b
        mse = np.mean((y_train - y_pred) ** 2)
        mse_history.append(mse)

        dw = (-2 / n) * np.sum(x_train_scaled * (y_train - y_pred))
        db = (-2 / n) * np.sum(y_train - y_pred)

        w -= lr * dw
        b -= lr * db

    return mse_history

epochs = 61
learning_rates = {
    0.001: ("#0E7C8B", "tốc độ học = 0.001"),
    0.1:   ("#1B7837", "tốc độ học = 0.1"),
    1.02:  ("#C02228", "tốc độ học = 1.02")
}

plt.figure(figsize=(7, 4.5))

for lr, (color, label) in learning_rates.items():
    mse_history = train_gd(lr, epochs)
    plt.plot(range(epochs), mse_history, color=color, linewidth=2, label=label)

plt.yscale("log")

plt.title("Tốc độ học quá nhỏ thì chậm, quá lớn thì hỏng", fontsize=12, fontweight="bold", pad=10)
plt.xlabel("Vòng lặp", fontsize=11, fontweight="bold")
plt.ylabel("MSE (thang loga)", fontsize=11, fontweight="bold")

plt.legend(loc="upper left", fontsize=9)
plt.grid(True, alpha=0.3, which="both", linestyle="-")
plt.tight_layout()

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv("gia_nha.csv")

np.random.seed(42)
if "so_phong" not in df.columns:
    df["so_phong"] = np.random.randint(1, 5, size=len(df))
if "khoang_cach" not in df.columns:
    df["khoang_cach"] = np.random.uniform(1, 15, size=len(df))

X_1var = df[["dien_tich"]]
X_3var = df[["dien_tich", "so_phong", "khoang_cach"]]
y = df["gia"].values

(X1_train, X1_test,
 X3_train, X3_test,
 y_train, y_test) = train_test_split(X_1var, X_3var, y, test_size=0.2, random_state=42)

model_1var = LinearRegression()
model_1var.fit(X1_train, y_train)
y_pred_1var = model_1var.predict(X1_test)
r2_1var = r2_score(y_test, y_pred_1var)

model_3var = LinearRegression()
model_3var.fit(X3_train, y_train)
y_pred_3var = model_3var.predict(X3_test)
r2_3var = r2_score(y_test, y_pred_3var)

plt.figure(figsize=(6, 6))

min_val = min(y_test.min(), y_pred_1var.min(), y_pred_3var.min()) - 0.5
max_val = max(y_test.max(), y_pred_1var.max(), y_pred_3var.max()) + 0.5
plt.plot([min_val, max_val], [min_val, max_val], color="#4A4A4A", linestyle="--", linewidth=2)

plt.scatter(
    y_test,
    y_pred_1var,
    color="#3868A0",
    s=45,
    label=rf"Một biến, \(R^2\) = {r2_1var:.4f}",
    zorder=3
)

plt.scatter(
    y_test,
    y_pred_3var,
    color="#C02228",
    marker="^",
    s=50,
    label=rf"Ba biến, \(R^2\) = {r2_3var:.4f}",
    zorder=4
)

plt.title("Thêm biến, điểm bám sát đường chéo hơn", fontsize=12, fontweight="bold", pad=12)
plt.xlabel("Giá thật (tỷ đồng)", fontsize=11, fontweight="bold")
plt.ylabel("Giá dự đoán (tỷ đồng)", fontsize=11, fontweight="bold")

plt.xlim(min_val, max_val)
plt.ylim(min_val, max_val)
plt.legend(loc="upper left", fontsize=9)
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```
